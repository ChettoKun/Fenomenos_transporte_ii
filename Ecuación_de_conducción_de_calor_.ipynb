## **Ecuación unidimensional de la conducción de calor**

Considere la conducción de calor a través de una pared plana grande, como la de una casa, el vidrio de una ventana de una sola hoja, la placa metálica de la base de una plancha, un tubo para vapor de agua de hierro fundido, un elemento cilíndrico de combustible nuclear, una resistencia eléctrica de alambre, la pared de un recipiente esférico o una bola metálica que está siendo templada por inmersión o revenida. La conducción de calor en esta y muchas otras configuraciones geométricas se puede considerar unidimensional, ya que la conducción a través de ellas será dominante en una dirección y despreciable en las demás. Enseguida, se desarrollará la ecuación unidimensional de la conducción de calor en coordenadas rectangulares, cilíndricas y esféricas.

### **Ecuación de la conducción de calor en una pared plana grande**

Considere un elemento delgado de espesor $\Delta x$ en una pared plana grande, como se muestra en la siguiente figura.

imagen.png

Suponga que la densidad de la pared es $\rho$, el calor especifico es $C$ y el área de la pared perpendicular a la dirección de transferencia de calor es $A$. Un *balance de energía* sobre este elemento delgado, durante un pequeño intervalo de tiempo $\Delta t$, se puede expresar como:

$$\dot{Q}_x-\dot{Q}_{x+\Delta x}+\dot{E}_{gen,elemento}=\frac{\Delta E_{elemento}}{\Delta t}$$

Pero el cambio en el contenido de energía interna del elemento y la velocidad de generación de calor dentro del elemento se pueden expresar como:

$$\Delta E_{elemento}=E_{t+\Delta t}-E_t=mc(T_{t+\Delta t}-T_t)=\rho cA\Delta x(T_{t+\Delta t}-T_t)$$

$$\dot{E}_{gen,elemento}=\dot{e}V=\dot{e}_{gen}A\Delta x$$

Por lo tanto, se obtiene

$$\dot{Q}_x-\dot{Q}_{x+\Delta x}+\dot{e}A\Delta x=\rho cA\Delta x\frac{T_{t+\Delta t-T_t}}{\Delta t}$$

Al dividir entre $A\Delta x$ da
$$-\frac{1}{A}\frac{\dot{Q}_{x+\Delta x}-\dot{Q}_x}{\Delta x}+\dot{e}_{gen}=\rho c\frac{T_{t+\Delta t}-T_t}{\Delta t}$$

Al tomar el límite cuando $\Delta x \rightarrow 0$ y $\Delta t \rightarrow 0$, se obtiene la ecuación de conducción de calor con **conductividad variable**

$$\frac{\partial}{\partial{x}}\left(k\frac{\partial{T}}{\partial{x}} \right)+\frac{\dot{e}_{gen}}{k}=\rho c \frac{\partial{T}}{\partial{t}}$$

En general, la conductividad térmica $k$ de un material depende de la temperatura $T$ (y, por lo tanto, de $x$) y, por consiguiente, no se puede extraer de la derivada. No obstante, en la mayor parte de las aplicaciones prácticas se puede suponer que la *conductividad térmica* permanece *constante* en algún valor promedio. En ese caso, la ecuación de calor se reduce a:

$$\frac{\partial}{\partial{x}}\left(\frac{\partial{T}}{\partial{x}} \right)+\frac{\dot{e}_{gen}}{k}=\frac{1}{\alpha} \frac{\partial{T}}{\partial{t}}$$

donde la propiedad $\alpha=k/\rho c$ es la **difusividad térmica** del material y representa la velocidad con que se propaga el calor a través del mismo. Ésta se reduce a las formas siguientes en condiciones específicas:

**Régimen estacionario $\frac{\partial{T}}{\partial{t}}=0$**

$$\frac{d^2T}{dx^2}+\frac{\dot{e}_{gen}}{k}=0$$

**Régimen transitorio, sin generación de calor  $\dot{e}_{gen}=0$**

$$\frac{\partial^2{T}}{\partial{x}^2}=\frac{1}{\alpha}\frac{\partial{T}}{\partial{t}}$$

**Régimen estacionario, sin generación de calor $\partial{T}/\partial{t}=0, \dot{e}_{gen}$**

$$\frac{d^2T}{dx^2}=0$$

### **Ecuación de la conducción de calor en un cilindro largo**

Considere ahora un elemento delgado con forma de casco cilíndrico, de espesor $\Delta r$, en un cilindro largo, como se muestra en la siguiente figura.

imagen.png

Suponga que la densidad del cilindro es $\rho$, el calor específico es $c$ y la longitud es $L$. El área del cilindro, normal a la dirección de transferencia de calor en cualquier lugar, es $A=2\pi rL$, donde $r$ es el valor del radio en ese lugar. Note que el área $A$ de la transferencia de calor depende de $r$ en este caso y, por tanto, varía con el lugar. Un *balance de energía* sobre este elemento delgado con forma de casco cilíndrico, durante un pequeño intervalo de tiempo $\Delta t$, se puede expresar como:

$$\dot{Q}_r-\dot{Q}_{r+\Delta r}+\dot{E}_{gen, elemento}=\frac{\Delta E_{elemento}}{\Delta t}$$

El cambio en el contenido de energía del elemento y la velocidad de generación de calor dentro del mismo se puede expresar como:

$$\Delta E_{elemento}=E_{t+\Delta t}-E_t=mc(T_{T+\Delta t}-T_t)=\rho cA\Delta r(T_{t+\Delta t}-T_t)$$

$$\dot{E}_{gen,elemento}=\dot{e}_{gen}V_{elemento}=\dot{e}_{gen}A\Delta r$$

Por lo tanto,

$$\dot{Q}_r-\dot{Q}_{r+\Delta r}+\dot{e}A\Delta r=\rho A\Delta r\frac{T_{t+\Delta t}-T_t}{\Delta t}$$

donde $A=2\pi rL$. El lectro puede sentirse tentado a expresar el área localizada a la *mitad* del elemento, usando el radio *promedio* como $A=2\pi (r+\Delta r/2)L$. Pero nada hay que se pueda ganar a partir de esta complicación, ya que, más adelante en el análisis, se tomará el límite cuando $\Delta r \rightarrow 0$ y, por tanto, se cancelará el término $\Delta r/2$. Ahora se divide la ecuación anterior entre $A\Delta r$ y da

$$-\frac{1}{A}\frac{\dot{Q}_{r+\Delta r}-\dot{Q}_r}{\Delta r}+\dot{e}_{gen}=\rho c\frac{T_{t+\Delta t}-T_t}{\Delta t}$$

Si se toma el límite cuando $\Delta r \rightarrow 0$ y $\Delta t \rightarrow 0$, se obtiene

$$-\frac{1}{A}\frac{\partial}{\partial{r}}\left(kA\frac{\partial{T}}{\partial{r}}\right)+\dot{e}_{gen}=\rho c\frac{\partial{T}}{\partial{t}}$$

Puesto que el área de trasnferencia de calor en este caso es $A=2\pi rL$, la ecuación unidimensiona de conducción de calor en régimen transitorio en un cilindro queda

**Conductividad variable**

$$\frac{1}{r}\frac{\partial}{\partial{r}}\left(rk\frac{\partial{T}}{\partial{r}} \right)+\dot{e}_{gen}=\rho c\frac{\partial{T}}{\partial{t}}$$

**Conductividad constante**

$$\frac{1}{r}\frac{\partial}{\partial{r}}\left(r\frac{\partial{T}}{\partial{r}}  \right)+\frac{\dot{e}_{gen}}{k}=\frac{1}{\alpha}\frac{\partial{T}}{\partial{t}}$$

donde una vez más la propiedad $\alpha=k/\rho c$ es la **difusividad térmica** del material. En condiciones especificadas, la ecuación anterior se reduce a las formas siguientes:

**Régimen estacionario $(\partial{T}/\partial{t}=0)$**

$$\frac{1}{r}\frac{\partial}{\partial{r}}\left(r\frac{\partial{T}}{\partial{r}} \right)+ \frac{\dot{e}_{gen}}{k}=0$$

**Régimen transitorio, sin generación de calor  ($\dot{e}_{gen}=0$**

$$\frac{1}{r}\frac{\partial}{\partial{r}}\left( r\frac{\partial{T}}{\partial{r}} \right)=\frac{1}{\alpha}\frac{\partial{T}}{\partial{t}}$$

**Régimen estacionario, sin generación de calor ($(\partial{T}/\partial{t}=0)$ y $\dot{e}_{gen}=0$)**

$$\frac{d}{dr}\left(r\frac{dT}{dr} \right)=0$$

Note que, una vez más, se reemplazan las derivadas parciales por derivadas ordinarias en el caso de la conducción unidimensional y estacionara de calor, ya que son idénticas cuando dicha función depende de una sola variable ($T=T(r)$).

### **Ecuación de la conducción de calor en una esfera**

Considere ahora una esfera con densidad $\rho$, calor específico $c$ y radio exterior $R$. El área de la esfera normal a la dirección de trasnferencia de calor, en cualquier lugar, es $A=4\pi r^2$, en donde $r$ es el valor del radio en ese lugar. Note que, en este caso, el área de transferencia de calor $A$, depende de $r$ y, por tanto, varía con la ubicación. Al considerar un elemento con forma de casco esférico delgado de espesor $\Delta r$ y repetir el procedimiento descrito con anterioridad para el cilindro, usando $A=4\pi r^2$, la cuación unidimensional de conducción de calor en régimen transitorio para una esfera se determina que es:

imagen.png

**Conductividad variable**

$$\frac{1}{r^2}\frac{\partial}{\partial{r}}\left(r^2k\frac{\partial{T}}{\partial{r}}  \right)+\dot{e}_{gen}+\rho c\frac{\partial{T}}{\partial{t}}$$

la cual, en el caso de conductividad térmica constante, se reduce a

**Conductividad constante**

$$\frac{1}{r^2}\frac{\partial}{\partial{r}}\left(r^2\frac{\partial{T}}{\partial{r}}  \right)+\frac{\dot{e}_{gen}}{k}+\frac{1}{\alpha}\frac{\partial{T}}{\partial{t}}$$

en donde, a su vez, la propiedad $\alpha=k/\rho c$ es la difusividad térmica del material. En condiciones especificadas, se reduce a las formas siguientes:

**Régimen estacionario $(\partial{T}/\partial{t}=0)$**

$$\frac{1}{r^2}\frac{\partial}{\partial{r}}\left(r^2\frac{\partial{T}}{\partial{r}} \right)+ \frac{\dot{e}_{gen}}{k}=0$$

**Régimen transitorio, sin generación de calor  ($\dot{e}_{gen}=0$**

$$\frac{1}{r^2}\frac{\partial}{\partial{r}}\left( r^2\frac{\partial{T}}{\partial{r}} \right)=\frac{1}{\alpha}\frac{\partial{T}}{\partial{t}}$$

**Régimen estacionario, sin generación de calor ($(\partial{T}/\partial{t}=0)$ y $\dot{e}_{gen}=0$)**

$$\frac{d}{dr}\left(r^2\frac{dT}{dr} \right)=0$$

o

$$r\frac{d^2T}{dr^2}+
2\frac{dT}{dr}=0$$

donde, de nuevo, se reemplazan las derivadas parciales por derivadas ordinarias en el caso de conducción unidimensional y estacionaria de calor.

### **Ecuacion unidimensional combinada de la conducción de calor**

Un examen de las ecuaciones unidimensionales de conducción de calor en régimen transitorio, para la pared plana, el cilindro y la esfera, revela que las tres se pueden expresar en una forma compacta como

$$\frac{1}{r^n}\frac{\partial}{\partial{r}}\left( r^nk\frac{\partial{T}}{\partial{r}} \right)+\dot{e}_{gen}=\rho c\frac{\partial{T}}{\partial{t}}$$

donde $n=0$ para una pared plana, $n=1$ para un cilindro y $n=2$ para una esfera. En el caso de una pared plana se acostumbra reemplazar la variable $r$ por $x$. Esta ecuación se puede simplificar para los casos de régimen estacionario o sin generación de calor como se describe con anterioridad.

## **Ejemplo 2**

Se usa un calentador de resitencia de alambre de 2 $kW$, con conductividad térmica $k=$15 $W/m°C$, diámetro $D=$0.4 $cm$ y longitud $L=$50 $cm$, para calentar agua al sumergirlo en ella (ver figura)

imagen.png

Suponiendo que la variación de la conductividad térmica del alambre con la temperatura es despreciable, obtenga la ecuación diferencial que describe la variación de la temperatura en el alambre durante una operación estacionaria.

```python
#Responda aquí
from pde import *
import numpy as np

# Definir la malla en coordenadas cilíndricas (r, z)
grid = CylindricalSymGrid(radius=[0, 10], bounds_z=[0, 50], shape=(50, 50))
field = ScalarField(grid)

D=20 #cm
L=50 #cm
q= 2000 #W

vol=(np.pi*(D**2/2)*L) #cm^3
q_vol=q/vol #W/(cm^3), y es el e_puntito
k=15 #W/(cm*°C)

grid=CylindricalSymGrid(radius=10, bounds_z=(0,50), shape=[50,100])
field=ScalarField(grid, q_vol/k)
bc_r=[{"value":30},{"value":30}]
bc_z={"derivative":q_vol/k}
ecuacion=solve_poisson_equation(-field, bc=[bc_r, bc_z], label="Solución de la Ecuación de Poisson \nen coordenadas cilíndricas") #El negativo por el formato de la Ec. de Poisson
```

```python
ecuacion.plot(cmap="inferno")
```

## **Ejemplo 3**

Una esfera metálica de radio $R$ se calienta en un horno hasta la temperatura de 600 $°F$ en toda ella y, a continuación, se le extrae del horno y se deja enfriar en el aire ambiental a $T_∞$=75 $°F$, por convección y radiación (ver figura).

imagen.png

Se sabe que la conductividad térmica del material de la bola varía en forma lineal con al temperatura. Suponiendo que la bola se enfría uniformemente partiendo de toda la superficie exterior, obtenga la ecuación diferencia que describe la variación de la temperatura en la esfera durante el enfriamiento.

```python
#Responder aquí
grid=SphericalSymGrid(radius=[0, 20], shape=100)
field=ScalarField(grid,data=600)
bcr=[{"value":75}]

a=10
tracker=LivePlotTracker(interrupts=0.01,plot_args={"kind":"image"})
eq=PDE({"T":f"{a}*laplace(T)"},bc=bcr)

solucion=eq.solve(field, t_range=10, dt=0.00001,tracker=tracker)
```

## **Ejemplo 4**

Escriba y resuelva la ecuación de conducción del calor en dos dimensiones en régimen transitorio para una pared plana, con conductividad térmica constante ($k=40 W/cm °C$) y generación de calor ($\dot{e}=500 W/cm^3$). Indique qué representa cada una de las variables. Asuma una $\rho=1.4 g/cm^3$ y $c=0.8 J/g °C$. Condiciones de frontera $x=0, T_{x=0}=20 °C$, $x=L, T_{x=L}=30°C$, $y=0, T_{y=0}=60 °C$ e $y=L, T_{y=L}=90°C$. Condición inicial $t=0, T_{t=0}=25°C$.

Consirando coordenadas rectangulares, $k$=cte, existencia de $\dot{e}_{gen}$ y
en régimen transitorio, por lo tanto, la ecuación diferencial es:


```python
#Definimos la malla
grid=CartesianGrid([[0,10],[0,10]],shape=[100,100])
#Parámetros
k=40 #W/(cm*°C)
e=500 #W/(cm^3)
rho=1.4 #g/(cm^3)
c=0.8 #J/(g*°C)
alfa=k/(rho*c) #Difusividad térmica
#Definimos la condición inicial
field=ScalarField(grid, 25) #T en t=0
#Definimos las condiciones de frontera
bc_x=[{"value":20},{"value":30}]
bc_y=[{"value":60},{"value":90}]
#Definimos la ecuación
ecuacion=PDE({"T":f"(laplace(T)+{e}/{k})*{alfa}"},bc=[bc_x,bc_y])
```

```python
#resolvemos la ecuación
tracker=LivePlotTracker(interrupts=0.1, plot_args={"kind":"image","cmap":"inferno"})
solucion=ecuacion.solve(field, t_range=5, dt=0.000001, tracker=tracker)
```

## **Ejemplo 7**

Partiendo de un balance de energía sobre un elemento circular de volumen, deduzca la ecuación bidimensional de conducción de calor en régimen estacionario, con una conductividad térmica constante y con generación de calor. Considere $\dot{e}=5000$ $W/cm^3$ y una conductividad térmica $k=150$ $ W/cm °C$. El elemento de volumen se encuentra expuesto a una temperatura circundate de $45$ $°C$.

imagen.png

Considerando coordenadas cilíndricas ($n=1$), $k=cte$, hay generación de calor y el régimen es estacionario, la ecuación diferencial es:
$$\frac{1}{r^1}\frac{\partial }{\partial r}(r^{1}k\frac{\partial T}{\partial r})+\dot{e}=\rho c\frac{\partial T}{\partial t}$$

dado que esta en regimen estacionario $\frac{\partial T}{\partial t}=0$
$$\frac{k}{r}\frac{\partial }{\partial r}(r\frac{\partial T}{\partial r})+\dot{e}=0$$

$$\frac{1}{r}(r\frac{\partial^2{T}}{\partial^2{r}}+\frac{\partial{T}}{\partial{r}})+\frac{\dot{e}_{gen}}{k}=0$$

$$\frac{\partial^2{T}}{\partial{r^2}}+\frac{1}{r}\frac{\partial{T}}{\partial{r}}+\frac{\dot{e}}{k}=0$$

```python
grid=SphericalSymGrid(radius=10,shape=100)
e=500
k=150
field=ScalarField(grid,e/k)
bc={"value":45}
ecuacion=solve_poisson_equation(field,bc=bc)
ecuacion.plot(kind="image")
```
